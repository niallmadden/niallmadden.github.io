Write a funcation called `censor_vowels()` which takes as input a string `text`, and returns a string which is the same, except every lower-case vowel is replaced by a `*`. 

For example:
* `censor_vowels('This is appaling and outrageous')` should return `'Th*s *s *pp*l*ng *nd **tr*g***s'`

* `censor_vowels('I love Python')` should return `'I l*v* Pyth*n'`

```python
def censor_vowels(text):
    new = text
    for letter in 'aeiou':
        new = new.replace(letter, '*')
    return new   
```

```python
censor_vowels('Hello')
```

```python
assert censor_vowels('This is appaling and outrageous') == 'Th*s *s *pp*l*ng *nd **tr*g***s'
assert censor_vowels('I love Python') == 'I l*v* Pyth*n'
```
